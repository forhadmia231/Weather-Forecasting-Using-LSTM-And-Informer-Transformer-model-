```python
# =====================================================
# COMPLETE THESIS CODE: LSTM + INFORMER-STYLE TRANSFORMER
# =====================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from statsmodels.tsa.stattools import adfuller
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (
    LSTM, Dense, Dropout, Input,
    MultiHeadAttention, LayerNormalization,
    GlobalAveragePooling1D
)
from tensorflow.keras.callbacks import EarlyStopping

# =====================================================
# 1. LOAD DATASET
# =====================================================

df = pd.read_csv("/kaggle/input/datasets/mdforhadmia/marged-file/WEATHER_2021_2024_MERGED.csv")

print("Dataset Shape:", df.shape)
print(df.head())
print(df.columns)
print(df.isnull().sum())
print("Duplicate Rows:", df.duplicated().sum())

# =====================================================
# 2. DATE CREATE + SORT
# =====================================================

df["DATE"] = pd.to_datetime(df[["YEAR", "MONTH", "DAY"]])
df = df.sort_values("DATE")

features = ["TEMPERATURE", "HUMIDITY", "WIND_SPEED"]
data = df[features].copy()

# =====================================================
# 3. MISSING VALUE HANDLING
# =====================================================

data = data.interpolate(method="linear")
data = data.bfill()
data = data.ffill()

print("\nMissing Values After Cleaning:")
print(data.isnull().sum())

# =====================================================
# 4. DESCRIPTIVE STATISTICS TABLE
# =====================================================

desc_stats = data.describe().T
print("\nDescriptive Statistics:")
print(desc_stats)

desc_stats.to_csv("/kaggle/working/descriptive_statistics_table.csv")

# =====================================================
# 5. TREND GRAPHS
# =====================================================

for col in features:
    plt.figure(figsize=(15,5))
    plt.plot(data[col])
    plt.title(f"{col} Trend")
    plt.xlabel("Time")
    plt.ylabel(col)
    plt.grid(True)
    plt.savefig(f"/kaggle/working/{col}_trend.png", dpi=300, bbox_inches="tight")
    plt.show()

# =====================================================
# 6. CORRELATION MATRIX
# =====================================================

corr = data.corr()
print("\nCorrelation Matrix:")
print(corr)

corr.to_csv("/kaggle/working/correlation_matrix.csv")

plt.figure(figsize=(7,5))
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(features)), features, rotation=45)
plt.yticks(range(len(features)), features)

for i in range(len(features)):
    for j in range(len(features)):
        plt.text(j, i, round(corr.iloc[i, j], 2), ha="center", va="center")

plt.title("Correlation Matrix")
plt.savefig("/kaggle/working/correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()

# =====================================================
# 7. ADF TEST
# =====================================================

def adf_test(series, name):
    result = adfuller(series.dropna())
    status = "Stationary" if result[1] <= 0.05 else "Non-Stationary"
    return [name, result[0], result[1], status]

adf_results = []

for col in features:
    adf_results.append(adf_test(data[col], col))

adf_df = pd.DataFrame(
    adf_results,
    columns=["Variable", "ADF Statistic", "p-value", "Status"]
)

print("\nADF Test Results:")
print(adf_df)

adf_df.to_csv("/kaggle/working/adf_original_results.csv", index=False)

# Differencing ADF
diff_data = data.diff().dropna()

adf_diff_results = []

for col in features:
    adf_diff_results.append(adf_test(diff_data[col], col + "_DIFF"))

adf_diff_df = pd.DataFrame(
    adf_diff_results,
    columns=["Variable", "ADF Statistic", "p-value", "Status"]
)

print("\nADF Test After Differencing:")
print(adf_diff_df)

adf_diff_df.to_csv("/kaggle/working/adf_differenced_results.csv", index=False)

# =====================================================
# 8. NORMALIZATION
# =====================================================

scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)

scaled_df = pd.DataFrame(scaled_data, columns=features)

# =====================================================
# 9. SEQUENCE GENERATION
# =====================================================

sequence_length = 30

X = []
y = []

for i in range(sequence_length, len(scaled_df)):
    X.append(scaled_df.iloc[i-sequence_length:i].values)
    y.append(scaled_df.iloc[i].values)

X = np.array(X)
y = np.array(y)

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)

# =====================================================
# 10. TRAIN TEST SPLIT
# =====================================================

split = int(0.8 * len(X))

X_train = X[:split]
X_test = X[split:]

y_train = y[:split]
y_test = y[split:]

print("\nX_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

# Save split info
split_info = pd.DataFrame({
    "Item": ["Total Samples", "Training Samples", "Testing Samples", "Sequence Length", "Features"],
    "Value": [len(X), len(X_train), len(X_test), sequence_length, len(features)]
})

split_info.to_csv("/kaggle/working/train_test_split_table.csv", index=False)

# =====================================================
# 11. BUILD LSTM MODEL
# =====================================================

model_lstm = Sequential([
    LSTM(64, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])),
    Dropout(0.2),
    LSTM(32, return_sequences=False),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(3)
])

model_lstm.compile(
    optimizer="adam",
    loss="mse"
)

model_lstm.summary()

# =====================================================
# 12. TRAIN LSTM
# =====================================================

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history_lstm = model_lstm.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=64,
    validation_split=0.2,
    callbacks=[early_stop],
    verbose=1
)

# =====================================================
# 13. LSTM LOSS GRAPH
# =====================================================

plt.figure(figsize=(10,5))
plt.plot(history_lstm.history["loss"], label="Training Loss")
plt.plot(history_lstm.history["val_loss"], label="Validation Loss")
plt.title("LSTM Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.savefig("/kaggle/working/lstm_training_loss.png", dpi=300, bbox_inches="tight")
plt.show()

# =====================================================
# 14. LSTM PREDICTION + EVALUATION
# =====================================================

y_pred_lstm = model_lstm.predict(X_test)

y_test_actual = scaler.inverse_transform(y_test)
y_pred_lstm_actual = scaler.inverse_transform(y_pred_lstm)

lstm_results = []

for i, col in enumerate(features):
    mae = mean_absolute_error(y_test_actual[:, i], y_pred_lstm_actual[:, i])
    mse = mean_squared_error(y_test_actual[:, i], y_pred_lstm_actual[:, i])
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test_actual[:, i], y_pred_lstm_actual[:, i])

    lstm_results.append(["LSTM", col, mae, mse, rmse, r2])

lstm_results_df = pd.DataFrame(
    lstm_results,
    columns=["Model", "Variable", "MAE", "MSE", "RMSE", "R2 Score"]
)

print("\nLSTM Results:")
print(lstm_results_df)

lstm_results_df.to_csv("/kaggle/working/lstm_performance_results.csv", index=False)

# =====================================================
# 15. LSTM ACTUAL VS PREDICTED GRAPHS
# =====================================================

for i, col in enumerate(features):
    plt.figure(figsize=(15,5))
    plt.plot(y_test_actual[:300, i], label="Actual")
    plt.plot(y_pred_lstm_actual[:300, i], label="Predicted")
    plt.title(f"LSTM Actual vs Predicted: {col}")
    plt.xlabel("Time")
    plt.ylabel(col)
    plt.legend()
    plt.grid(True)
    plt.savefig(f"/kaggle/working/lstm_actual_vs_predicted_{col}.png", dpi=300, bbox_inches="tight")
    plt.show()

# =====================================================
# 16. BUILD INFORMER-STYLE TRANSFORMER MODEL
# =====================================================

def build_informer_model(input_shape, output_dim):
    inputs = Input(shape=input_shape)

    x = Dense(64)(inputs)

    attn_output = MultiHeadAttention(
        num_heads=4,
        key_dim=16
    )(x, x)

    attn_output = Dropout(0.2)(attn_output)
    x = LayerNormalization()(x + attn_output)

    ffn = Dense(128, activation="relu")(x)
    ffn = Dropout(0.2)(ffn)
    ffn = Dense(64)(ffn)

    x = LayerNormalization()(x + ffn)

    attn_output = MultiHeadAttention(
        num_heads=4,
        key_dim=16
    )(x, x)

    attn_output = Dropout(0.2)(attn_output)
    x = LayerNormalization()(x + attn_output)

    x = GlobalAveragePooling1D()(x)

    x = Dense(64, activation="relu")(x)
    x = Dropout(0.2)(x)

    outputs = Dense(output_dim)(x)

    model = Model(inputs, outputs)

    model.compile(
        optimizer="adam",
        loss="mse"
    )

    return model

model_informer = build_informer_model(
    input_shape=(X_train.shape[1], X_train.shape[2]),
    output_dim=3
)

model_informer.summary()

# =====================================================
# 17. TRAIN INFORMER
# =====================================================

history_informer = model_informer.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=64,
    validation_split=0.2,
    callbacks=[early_stop],
    verbose=1
)

# =====================================================
# 18. INFORMER LOSS GRAPH
# =====================================================

plt.figure(figsize=(10,5))
plt.plot(history_informer.history["loss"], label="Training Loss")
plt.plot(history_informer.history["val_loss"], label="Validation Loss")
plt.title("Informer Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.savefig("/kaggle/working/informer_training_loss.png", dpi=300, bbox_inches="tight")
plt.show()

# =====================================================
# 19. INFORMER PREDICTION + EVALUATION
# =====================================================

y_pred_informer = model_informer.predict(X_test)
y_pred_informer_actual = scaler.inverse_transform(y_pred_informer)

informer_results = []

for i, col in enumerate(features):
    mae = mean_absolute_error(y_test_actual[:, i], y_pred_informer_actual[:, i])
    mse = mean_squared_error(y_test_actual[:, i], y_pred_informer_actual[:, i])
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test_actual[:, i], y_pred_informer_actual[:, i])

    informer_results.append(["Informer", col, mae, mse, rmse, r2])

informer_results_df = pd.DataFrame(
    informer_results,
    columns=["Model", "Variable", "MAE", "MSE", "RMSE", "R2 Score"]
)

print("\nInformer Results:")
print(informer_results_df)

informer_results_df.to_csv("/kaggle/working/informer_performance_results.csv", index=False)

# =====================================================
# 20. INFORMER ACTUAL VS PREDICTED GRAPHS
# =====================================================

for i, col in enumerate(features):
    plt.figure(figsize=(15,5))
    plt.plot(y_test_actual[:300, i], label="Actual")
    plt.plot(y_pred_informer_actual[:300, i], label="Predicted")
    plt.title(f"Informer Actual vs Predicted: {col}")
    plt.xlabel("Time")
    plt.ylabel(col)
    plt.legend()
    plt.grid(True)
    plt.savefig(f"/kaggle/working/informer_actual_vs_predicted_{col}.png", dpi=300, bbox_inches="tight")
    plt.show()

# =====================================================
# 21. COMPARATIVE PERFORMANCE TABLE
# =====================================================

comparison_df = pd.concat(
    [lstm_results_df, informer_results_df],
    ignore_index=True
)

print("\nComparative Performance:")
print(comparison_df)

comparison_df.to_csv("/kaggle/working/comparative_performance_table.csv", index=False)

# =====================================================
# 22. RMSE COMPARISON BAR CHART
# =====================================================

plt.figure(figsize=(10,5))

for model_name in comparison_df["Model"].unique():
    temp_df = comparison_df[comparison_df["Model"] == model_name]
    plt.bar(
        temp_df["Variable"] + " (" + model_name + ")",
        temp_df["RMSE"]
    )

plt.title("RMSE Comparison Between LSTM and Informer")
plt.xlabel("Variable")
plt.ylabel("RMSE")
plt.xticks(rotation=45)
plt.grid(True)
plt.savefig("/kaggle/working/rmse_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

# =====================================================
# 23. R2 COMPARISON BAR CHART
# =====================================================

plt.figure(figsize=(10,5))

for model_name in comparison_df["Model"].unique():
    temp_df = comparison_df[comparison_df["Model"] == model_name]
    plt.bar(
        temp_df["Variable"] + " (" + model_name + ")",
        temp_df["R2 Score"]
    )

plt.title("R² Score Comparison Between LSTM and Informer")
plt.xlabel("Variable")
plt.ylabel("R² Score")
plt.xticks(rotation=45)
plt.grid(True)
plt.savefig("/kaggle/working/r2_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

# =====================================================
# 24. HYPERPARAMETER TABLE
# =====================================================

hyperparameter_table = pd.DataFrame({
    "Parameter": [
        "Input Features",
        "Sequence Length",
        "Train-Test Split",
        "Validation Split",
        "Epochs",
        "Batch Size",
        "Optimizer",
        "Loss Function",
        "Dropout Rate",
        "Early Stopping",
        "Early Stopping Patience",
        "Output Variables"
    ],
    "LSTM Model": [
        "Temperature, Humidity, Wind Speed",
        "30 days",
        "80:20",
        "20% of training data",
        "50",
        "64",
        "Adam",
        "Mean Squared Error",
        "0.20",
        "Yes",
        "10",
        "3"
    ],
    "Informer-Style Transformer": [
        "Temperature, Humidity, Wind Speed",
        "30 days",
        "80:20",
        "20% of training data",
        "50",
        "64",
        "Adam",
        "Mean Squared Error",
        "0.20",
        "Yes",
        "10",
        "3"
    ]
})

print("\nHyperparameter Table:")
print(hyperparameter_table)

hyperparameter_table.to_csv("/kaggle/working/hyperparameter_table.csv", index=False)

# =====================================================
# 25. MODEL ARCHITECTURE TABLES
# =====================================================

lstm_architecture = pd.DataFrame({
    "Layer": [
        "Input Layer",
        "LSTM Layer 1",
        "Dropout 1",
        "LSTM Layer 2",
        "Dropout 2",
        "Dense Layer",
        "Output Layer"
    ],
    "Configuration": [
        "Input shape = 30 × 3",
        "64 units, return_sequences=True",
        "Dropout rate = 0.20",
        "32 units, return_sequences=False",
        "Dropout rate = 0.20",
        "32 units, ReLU activation",
        "3 neurons"
    ]
})

informer_architecture = pd.DataFrame({
    "Layer": [
        "Input Layer",
        "Input Projection",
        "Multi-Head Attention Block 1",
        "Dropout",
        "Layer Normalization",
        "Feed Forward Network",
        "Multi-Head Attention Block 2",
        "Global Average Pooling",
        "Dense Layer",
        "Dropout",
        "Output Layer"
    ],
    "Configuration": [
        "Input shape = 30 × 3",
        "Dense layer with 64 units",
        "4 heads, key_dim = 16",
        "Dropout rate = 0.20",
        "Residual connection + normalization",
        "Dense 128 ReLU → Dropout 0.20 → Dense 64",
        "4 heads, key_dim = 16",
        "GlobalAveragePooling1D",
        "64 units, ReLU activation",
        "Dropout rate = 0.20",
        "3 neurons"
    ]
})

lstm_architecture.to_csv("/kaggle/working/lstm_architecture_table.csv", index=False)
informer_architecture.to_csv("/kaggle/working/informer_architecture_table.csv",index=False)

print("\nLSTM Architecture Table:")
print(lstm_architecture)

print("\nInformer Architecture Table:")
print(informer_architecture)

# =====================================================
# 26. SAVE FINAL PREDICTIONS
# =====================================================

prediction_df = pd.DataFrame({
    "Actual_Temperature": y_test_actual[:, 0],
    "LSTM_Predicted_Temperature": y_pred_lstm_actual[:, 0],
    "Informer_Predicted_Temperature": y_pred_informer_actual[:, 0],

    "Actual_Humidity": y_test_actual[:, 1],
    "LSTM_Predicted_Humidity": y_pred_lstm_actual[:, 1],
    "Informer_Predicted_Humidity": y_pred_informer_actual[:, 1],

    "Actual_Wind_Speed": y_test_actual[:, 2],
    "LSTM_Predicted_Wind_Speed": y_pred_lstm_actual[:, 2],
    "Informer_Predicted_Wind_Speed": y_pred_informer_actual[:, 2],
})

prediction_df.to_csv("/kaggle/working/final_actual_vs_predicted_values.csv", index=False)

# =====================================================
# 27. SAVE MODELS
# =====================================================

model_lstm.save("/kaggle/working/lstm_weather_model.h5")
model_informer.save("/kaggle/working/informer_weather_model.h5")

print("\n✅ ALL OUTPUTS SAVED SUCCESSFULLY!")
print("Check Kaggle /working folder for tables, graphs, predictions, and models.")
```
